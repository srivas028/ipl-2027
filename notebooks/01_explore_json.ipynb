```python
def match_summary(path):
    with open(path) as f:
        info = json.load(f)["info"]
    outcome = info["outcome"]
    if "winner" in outcome:
        by = outcome["by"]
        margin = f"{by['runs']} runs" if "runs" in by else f"{by['wickets']} wickets"
        result = f"{outcome['winner']} won by {margin}"
    else:
        result = outcome.get("result", "unknown")
    return {
        "date": info["dates"][0],
        "teams": " v ".join(info["teams"]),
        "venue": info.get("venue"),
        "toss": f"{info['toss']['winner']} chose to {info['toss']['decision']}",
        "result": result,
        "player_of_match": ", ".join(info.get("player_of_match", [])),
        "players": info.get("players"),
        "season": info.get("season")
    }
match_summary("../data/raw/ipl_json/335982.json")
```
